```python
df["TotalScore"] = df["Math"] + df["English"] + df["Science"]
df
```

```python
df ["AverageScore"] = (df["Math"] + df["English"] + df["Science"])/3
df
```

```python
sorted_df = df.sort_values("AverageScore",ascending=False,)
print(f"maximum three AverageScore\n{sorted_df.head(3)}")
```

```python
df.groupby("Gender",group_keys=True)["AverageScore"].mean()
```

```python
# Students who study 3 hours or less
low_study = df[df["StudyHours"] <= 3]

# Students who study more than 3 hours
high_study = df[df["StudyHours"] > 3]

print("Average score for 3 hours or less:")
print(low_study["AverageScore"].mean())

print("Average score for more than 3 hours:")
print(high_study["AverageScore"].mean())
```
